## Scraping Aggregation Page, Research Question : Do shows with more or fewer episodes get better ratings? 

```python
#Scraping Aggregation Page 
#Get request 
import requests
import time 
import re 
from bs4 import BeautifulSoup 

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36"
}
page = requests.get("https://mydramalist.com/shows/top", headers=headers, timeout=10)
page.raise_for_status()  
#Checking status code 
print(page.status_code)
print(page.text[:500])
```

```python
#Parses the html content using BeautifulSoup and returns a series of show dictionaries 
def get_shows(response):
    soup = BeautifulSoup(response.text, "html.parser")
    shows = []
    container_class = "box"

    containers = soup.find_all("div", class_=container_class)
    print(f"Found {len(containers)} show containers using class '{container_class}'")

    for item in containers:
        title_wrapper = item.find(class_="title")
        title_tag = title_wrapper.find("a") if title_wrapper else None

      
        info_tag = item.find(class_="text-muted")

        title = title_tag.text.strip() if title_tag else None
        url = title_tag["href"] if title_tag and title_tag.has_attr("href") else None
        info = info_tag.text.strip() if info_tag else None
        episodes = None
        #extracting number of episodes 
        if info:
            match = re.search(r"(\d+)\s+episodes?", info)
            if match:
                episodes = int(match.group(1))

        shows.append({
            "title": title,
            "url": url,
            "info": info,
            "episodes": episodes,
        })

    return shows
```

```python
if __name__ == "__main__":
    shows = get_shows(page)

    print(f"\nScraped {len(shows)} shows total.\n")

    for show in shows[:23]:
        print(show)
```
